# Generate Catalog

```python
import os
import pandas as pd
import numpy as np
```

## Initilize Directory

```python
model = ["cnn", "sta_lta"][1]
location = ["lunar", "mars"][0]
set = ["train", "test"][0]

test_set = [
    "S12_GradeB",
    "S15_GradeA",
    "S15_GradeB",
    "S16_GradeA",
    "S16_GradeB",
][4]

if set != "test" or location != "lunar":
    dir = f"./{model}/{location}/{set}/"
else:
    dir = f"./{model}/{location}/{set}/{test_set}/"

print(dir)
```

## Generate

```python
filename_col = []
time_abs_col = []
time_rel_col = []
velocity_col = []
true_col = []
prediction_col = []


files = [file for file in os.listdir(dir) if file.endswith(".csv")]

print(files)
```

```python
# if set != "test" or location != "lunar":
for file in files:
    df = pd.read_csv(os.path.join(dir, file))
    # print(df)

    length = len(df)
    # print("Length of data:          ", length)

    filename = str(file).split(".csv", 1)[0]
    filename_col.append(filename)
    # print("Filename:                ", filename[0])

    if set == "train":
        true = np.nonzero(df["true_arrival_time"])[0][0]
        true_col.append(true)
        # print("True Arrival Time:       ", true)

    prediction = np.nonzero(df["predicted_arrival_time"])[0][0]
    prediction_col.append(prediction)
    # print("Predicted Arrival Time:  ", prediction)

    time_abs = df["time_abs(%Y-%m-%dT%H:%M:%S.%f)"][prediction]
    time_abs_col.append(time_abs)
    # print("Absolute Time:           ", time_abs)

    time_rel = df["time_rel(sec)"][prediction]
    time_rel_col.append(time_rel)
    # print("Relative Time:           ", time_rel)

    velocity = df["velocity(m/s)"][prediction]
    velocity_col.append(velocity)
    # print("Velocity:                ", velocity)

print(filename_col)
print(time_abs_col)
print(time_rel_col)
print(velocity_col)
print(prediction_col)

if set == "train":
    results = pd.DataFrame(
        {
            "filename": filename_col,
            "time_abs(%Y-%m-%dT%H:%M:%S.%f)": time_abs_col,
            "time_rel(sec)": time_rel_col,
            "velocity(m/s)": velocity_col,
            "true_arrival_time": true_col,
            "predicted_arrival_time": prediction_col,
        },
    )
else:
    results = pd.DataFrame(
        {
            "filename": filename_col,
            "time_abs(%Y-%m-%dT%H:%M:%S.%f)": time_abs_col,
            "time_rel(sec)": time_rel_col,
            "velocity(m/s)": velocity_col,
            "predicted_arrival_time": prediction_col,
        }
    )

if set != "test" or location != "lunar":
    results.to_csv(f"./predictions_catalog__{model}_{location}_{set}.csv", index=False)

else:
    results.to_csv(
        f"./predictions_catalog__{model}_{location}_{set}_{test_set}.csv", index=False
    )
```
